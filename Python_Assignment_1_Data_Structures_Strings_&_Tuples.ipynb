# 1. String Concatenation & User Input Handling


```python
string1 = "Hello"
string2 = input("Enter your Name: ")
```

```python
concatenated_string = string1 + " " + string2
print(concatenated_string)
```

```python
string3 = concatenated_string + ", welcome to Python programming"
print(string3)
```

# 2. String Slicing and Indexing


```python
print("First character:", string3[0])
```

```python
print("Last character:", string3[-1])
```

```python
print("First 5 characters:", string3[:5])
```

```python
print("Last 11 characters:", string3[-11:])
```

```python
print("Reversed string:", string3[::-1])
```

```python
python_index = string3.find("Python")
if python_index != -1:
    print("Extracted word:", string3[python_index:python_index+6])
```

# 3. String Methods

```python
strM = "Python beginner tutorial."
```

```python
print("Uppercase:", strM.upper())
```

```python
print("Lowercase:", strM.lower())
```

```python
print("Capitalized:", strM.capitalize())
```

```python
print("Count of 't':", strM.count('t'))
```

```python
print("Replaced string:", strM.replace("Python", "Data Analytics"))
```

# 4. Tuples (Creation, Modification, and Access)

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
```

```python
t_combine = tuple1 + tuple2
print("t_combine:", t_combine)
```

```python
t_repeated = t_combine * 3
print("Repeated 3 times:", t_repeated)
```

```python
print("3rd element:", t_combine[2])
```

```python
print("First three elements:", t_combine[:3])
```

```python
print("Last three elements:", t_combine[-3:])
```
